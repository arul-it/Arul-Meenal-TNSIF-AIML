```python
# 1. Reverse an Array

arr = [10, 20, 30, 40, 50]

for i in range(len(arr) - 1, -1, -1):
    print(arr[i], end=" ")
```

```python
# 2. Count Even, Odd and Zero

arr = [10, 5, 0, 7, 8, 0, 13, 4]

even = 0
odd = 0
zero = 0

for num in arr:
    if num == 0:
        zero += 1
    elif num % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even:", even)
print("Odd:", odd)
print("Zero:", zero)
```

```python
# 3. Sum of Positive and Negative Numbers

arr = [10, -5, 20, -8, 15, -2]

positive_sum = 0
negative_sum = 0

for num in arr:
    if num > 0:
        positive_sum += num
    else:
        negative_sum += num

print("Positive Sum:", positive_sum)
print("Negative Sum:", negative_sum)
```

```python
# 4. Remove Duplicate Elements

arr = [10, 20, 10, 30, 20, 40, 30]

result = []

for num in arr:
    if num not in result:
        result.append(num)

print(result)
```

```python
# 5. Find the Missing Number

arr = [1, 2, 3, 5, 6, 7]
n = 7

total = n * (n + 1) // 2
sum_arr = 0

for num in arr:
    sum_arr += num

missing = total - sum_arr

print("Missing Number:", missing)
```

```python
# 6. Rotate an Array

arr = [1, 2, 3, 4, 5]
k = 2

k = k % len(arr)

result = arr[-k:] + arr[:-k]

print(result)
```

```python
# 7. Find the Most Frequent Element

arr = [2, 5, 2, 8, 5, 2, 3, 5, 2]

max_count = 0
most_frequent = arr[0]

for num in arr:
    count = 0

    for x in arr:
        if num == x:
            count += 1

    if count > max_count:
        max_count = count
        most_frequent = num

print("Most Frequent Element:", most_frequent)
print("Frequency:", max_count)
```

```python
# 8. Maximum Subarray Sum

arr = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

max_sum = arr[0]
current_sum = arr[0]

start = 0
best_start = 0
best_end = 0

for i in range(1, len(arr)):
    if arr[i] > current_sum + arr[i]:
        current_sum = arr[i]
        start = i
    else:
        current_sum += arr[i]

    if current_sum > max_sum:
        max_sum = current_sum
        best_start = start
        best_end = i

print("Maximum Subarray Sum:", max_sum)
print("Subarray:", arr[best_start:best_end + 1])
```

```python
# 9. Check Whether Two Arrays are Equal

arr1 = [1, 2, 2, 3, 4]
arr2 = [4, 2, 1, 2, 3]

if sorted(arr1) == sorted(arr2):
    print("Arrays are Equal")
else:
    print("Arrays are Not Equal")
```
